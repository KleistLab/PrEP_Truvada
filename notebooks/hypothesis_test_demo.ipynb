```python
# ---------------------------------------------------------------------
# Hypothesis-driven efficacy distributions
#
# Each element in `phi_` corresponds to a mechanistic hypothesis.
#
# For each hypothesis, we use a virtual population of 1000 individuals,
# where each individual has a time-averaged efficacy φ ∈ [0,1]
# computed from mechanistic PK/PD simulations.
#
# These φ values are sampled according to trial-specific detection/adherence probabilities.
# ---------------------------------------------------------------------

phi_ = [
    # Base hypothesis:
    # Efficacy arises solely from systemic drug exposure
    # (no local tissue contribution)
    np.load('../data/processed/hypothesis_driven/mean_base.npy'),

    # dNTP-only hypothesis:
    # Protection mediated by intracellular metabolite dynamics
    np.load('../data/processed/hypothesis_driven/mean_dntp.npy'),

    # PK-only hypothesis:
    # Protection determined by pharmacokinetics alone
    np.load('../data/processed/hypothesis_driven/mean_pk.npy'),

    # Combined PK + dNTP hypothesis
    np.load('../data/processed/hypothesis_driven/mean_combo.npy'),

    # Mixed systemic + local tissue protection (50% local contribution)
    np.load('../data/processed/hypothesis_driven/mean_virtual_50.npy'),

    # Mixed systemic + local tissue protection (30% local contribution)
    np.load('../data/processed/hypothesis_driven/mean_virtual_30.npy'),

    # Mixed systemic + local tissue protection (10% local contribution)
    np.load('../data/processed/hypothesis_driven/mean_virtual_10.npy'),
]
phi_hypotheses = {
    'Systemic PK': phi_[0],
    'Local dNTP': phi_[1],
    'Local PK': phi_[2],
    'Local PK and dNTP': phi_[3],
    '50 local PK': phi_[4],
    '30 local PK': phi_[5],
    '10 local PK': phi_[6],
}


# ---------------------------------------------------------------------
# Data-driven PrEP efficacy distributions
#
# Each element in `phi_topdown` is a posterior distribution
# P(φ | Inf_obs) inferred directly from clinical trial data
# using Bayesian inference.
#
# These distributions correspond to Eq. (1)–(4) in the Methods:
#
#   P(φ | Inf_obs) ∝ P(Inf_obs | φ) · P(φ)
#
# with:
#   - uniform prior P(φ) = U(0,1)
#   - binomial likelihood with infection probability
#     r_inf(φ) = r_inf(∅) · (1 − φ)
#
# These posteriors represent population-level efficacy estimates
# for individuals with detectable drug.
# ---------------------------------------------------------------------

phi_topdown = [
    # IPERGAY posterior efficacy distribution
    np.load('../data/processed/data_driven/phi_IPERGAY.npy'),

    # iPrEx posterior efficacy distribution
    np.load('../data/processed/data_driven/phi_iPrEX.npy'),

    # HPTN 083 posterior efficacy distribution
    np.load('../data/processed/data_driven/phi_HPTN083.npy'),

    # PURPOSE II posterior efficacy distribution
    np.load('../data/processed/data_driven/phi_PURPOSEII.npy'),

    # DISCOVER posterior efficacy distribution
    np.load('../data/processed/data_driven/phi_DISCOVER.npy'),
]
```

```python

trial_names = ['IPERGAY', 'iPrEx', 'HPTN083', 'PURPOSEII', 'DISCOVER']
p_sampling = [ipergay_pDet_001, iprex_pDet_035, hptn083_pAdh, purpose2_pAdh, discover_pAdh]

hyp_list = ['Systemic PK', 'Local dNTP','Local PK', 'Local PK and dNTP','50 local PK', '30 local PK','10 local PK']

res = run_hypotheses(
    trialname="HPTN083",
    hypotheses="all",
    nsim=100000,
    trial_names=trial_names,
    hyp_list=hyp_list,
    p_sampling=p_sampling,
    phi_topdown=phi_topdown,
    phi_=phi_hypotheses,
    trial_parameters_drug_det_men=trial_parameters_drug_det_men,
    trial_parameters_drug_undet_men=trial_parameters_drug_undet_men 
)
```
